#### Program checks if GPA of a student for meets Dean's List and Honor Roll

```python
while True:
    
    student_last_name = input("Please enter student's last name: ")

    if student_last_name == 'ZZZ':
        break
    student_first_name = input("Please enter student's first name: ")
    student_gpa = float(input("Enter student GPA: "))
    
    if student_gpa >= 3.5:
        print(f"{student_first_name} {student_last_name} has made the Dean's List.")
    if student_gpa >= 3.25:
        print(f"{student_first_name} {student_last_name} has made the Honor Roll.")
    else:
        print("This student has not made the Dean's List or Honor Roll.")
        
        

```
